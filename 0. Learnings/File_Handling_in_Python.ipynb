##**How does open() works?**

1. Python calls the operating system's file-opening system call
2. The OS checks:
whether the file exists,
permissions,
location on disk
* gives Python a connection/handle to that file
3. Python creates a file object (f) so you can easily use commands like:
read()
write()
close()
4. while reading/writing, Python transfers data between program and the file through that connection.
5. Python also uses a temporary memory area called a buffer to make file operations faster instead of accessing the disk every single time.

##**Python supports various file modes:**

* 'r' – Read (default mode)
* 'w' – Write (overwrites if file exists)
* 'a' – Append (adds data to the file)
* 'x' – Create (fails if file exists)
* 'b' – Binary mode (e.g., 'rb', 'wb')
* 't' – Text mode (default, e.g., 'rt', 'wt')
* '+' – Read and write (e.g., 'r+', 'w+', 'a+')

```python
#Problem with w mode - replacing of data
#append mode

f= open('temp.txt', 'a')
f.write('\nAppended text')
f.close()
```

```python
#writing multiple lines in the file as a list

L = ['this\n', 'is\n', 'the\n', 'text\n', 'added\n', 'as a single list\n']

f = open('temp.txt', 'w')
f.writelines(L)
f.close()
```

```python
#reading from files - using read()

f = open('temp.txt', 'r')
g = f.read()
print(g)
f.close()
```

```python
#reading upto n chars

f = open('temp.txt', 'r')
g = f.read(10)
print(g)
f.close()
```

```python
#readin file line by line - readLine()

f = open('temp.txt', 'r')
print(f.readline())
print(f.readline())
f.close()

#readline() changes the line automatically and so does print(). therefore, a line will be left empty between the actual lines
```

```python
#reading entire file using readline()

f = open('temp.txt', 'r')

while True:
  data = f.readline()

  if data == '':
    break
  else:
    print(data, end = '')

f.close()
```

* read(size): Reads size characters (or whole file if not specified).
* readline(): Reads a single line from the file.
* readlines(): Reads all lines and returns a list.

##**Using Content Manager**

* close() is used to close the file to free up reasources, However if we won't, garbage collector will.
* 'with' keyword closes the file as soon as the usage is over.

```python
with open('temp.txt', 'a') as f:
  f.write('this is the text using \'with\' keyword')
```

```python
#reading using 'with'

with open('temp.txt', 'r') as f:
  print(f.read())
```

```python
#moving within a file -> 10 chars then 10 chars, lets say

with open('temp.txt', 'r') as f:
  print(f.read(10))
  print(f.read(10))
```

```python
#benefit - loading big files

L = ['hello ' for i in range(1000)]

with open('big.txt', 'w') as f:
  f.writelines(L)

```

```python
with open('big.txt', 'r') as f:
  chunk_size = 100
  while len(f.read(chunk_size)) > 0:
    print(f.read(chunk_size))
    f.read(chunk_size)
```

```python
#seek and tell functions

with open('temp.txt', 'r') as f:
  print(f.read(10))
  print(f.tell()) #tells where the pointer is
  f.seek(0) #moves the pointer to the given position
  print(f.read(10))
```

##**Serialization and Deserialization**


* Serialization : process of converting python data type to JSON format
* Deserializtion : process of converting JSON to python data types

JSON (JavaScript Object Notation)


```python
#serialization usign JSON

import json

l = [1,2,3,4]
with open('demo.json', 'w') as f:
  json.dump(l, f)
```

```python

d = {'name': 'khushi', 'age' : 22, 'gender': 'female'}

with open('demo.json', 'w') as f:
  json.dump(d, f, indent = 4)
```

```python
#deserialization

import json

with open('demo.json', 'r') as f:
  d = json.load(f)
  print(d)
  print(type(d))
```

###**Serializing and deserializing custom objects**



```python
class Person:
  def __init__(self, fname, lname, age, gender):
    self.fname = fname
    self.lname = lname
    self.age = age
    self.gender = gender
#format should be :
# -> Khushi rawat age -> 22 gender -> female
```

```python
person = Person('Khushi', 'Rawat', 22, 'female')
```

```python
import json

#with open('demo.json', 'w') as f:
#  json.dump(person, f)     #Object of type Person is not JSON serializable


def showObject(person):
  if isinstance(person, Person):
    return {'name':person.fname + ' ' + person.lname, 'age': person.age, 'gender': person.gender}

with open('demo.json', 'w') as f:
  json.dump(person, f, default =  showObject, indent =4)
```

```python
#deserialization

import json

with open('demo.json', 'r') as f:
  d = json.load(f)
  print(d)
  print(type(d))
```

##**Pickling**


Pickling : process of converting python objects intro byte stream (binary file)
Unpickling : process of converting back the byte stream (from a binary file) back to the object heirarchy


```python
class Person2:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def display(self):
    print("My name is ", self.name, " and my age is : ", self.age)
```

```python
p = Person2('khushi', 22)
```

```python
#pickle dump

import pickle
with open('person.pkl', 'wb') as f:
  pickle.dump(p, f)
```

```python
#pickle load
import pickle

with open('person.pkl', 'rb') as f:
  r = pickle.load(f)

r.display()
```

##**Pickle vs JSON**

* Pickle lets the user to store data in binary format.
* JSON lets the user to store data in human readable format.
