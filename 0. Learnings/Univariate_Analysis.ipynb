```python
import pandas as pd
import seaborn as sns
```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)
```

```python
df.head()
```

##**Caterogorical**

###**1. Count Plot**

```python
sns.countplot(df['Embarked'])

#same as
#df['Embarked'].value_counts().plot(kind = 'bar')
```

###**2. pie chart**

```python
df['Sex'].value_counts().plot(kind = 'pie', autopct = '%.2f')
#there is no direct alternative for a pie chart in seaborn
```

##**Categorical Data**

###**1. Historgram**

```python
import matplotlib.pyplot as plt
```

```python
#plt.hist(df['Age'])
plt.hist(df['Age'], bins = 20)
```

###**2. Distplot**

```python
sns.distplot(df['Age'])  #deprecated
```

```python
sns.displot(df['Age'], kde = True) #for newer versions
```

###**3. BoxPlot**

```python
sns.boxplot(df['Age'])
```
