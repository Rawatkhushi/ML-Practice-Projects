```python
class GDRegressor:

    def __init__(self, lr = 0.01, epochs = 100):

        self.coef_ = None
        self.intercept = None
        self.epochs = epochs
        self.lr= lr

    def fit(self, X_train, y_train):

        self.intercept = 0
        self.coef_ = np.ones(X_train.shape[1])

        for i in range(self.epochs):
            y_hat = np.dot(X_train, self.coef_) +  self.intercept
            intercept_derivative = -2 * np.mean(y_train - y_hat)
            self.intercept= self.intercept - (self.lr * intercept_derivative) 

            coef_derivative = -2 * np.dot((y_train - y_hat),X_train)/X_train.shape[0]
            self.coef_ = self.coef_ - (self.lr * coef_derivative)
        print(self.intercept, self.coef_)
        
    def predict(self, X_test):
        return np.dot(X_test, self.coef_) + self.intercept
        
```

```python
gdr = GDRegressor(epochs=1000,lr=0.5)
gdr.fit(X_train,y_train)
```

```python
y_pred = gdr.predict(X_test)
r2_score(y_test,y_pred)
```
