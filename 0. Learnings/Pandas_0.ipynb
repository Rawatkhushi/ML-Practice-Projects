#**Pandas**

###**What**
* python library works with structured data - table, csv files, spreadsheet

###**How**
* used in cleaning/analyzing the dataset before training ML model on it.

###**Why**
* because the data can be inconsistent. For ex,
1. missing values
2. duplicate rows
3. improper format
4. textual values mixed with strings

###**Alternatives**
* Polars : faster than pandas, lower memory usage, built on **Rust**
* PySpark : industry standard, less beginner friendly, distributed clusters

###**Data Structure in Pandas**

####**Series** : 1d array of any data type

####**DataFram** : a 2d (rows and columns) structure

```python
import pandas as pd
```

###**Series from a List**

```python
#numbers
marks = [1, 2]

new = pd.Series(marks)

#strings
name = ['khushi', 'rawat']
newser = pd.Series(name)
```

```python
#custom index

pd.Series(marks, index = name)
```

```python
#setting a name

pd.Series(marks, name = 'Khushi Rawat')
```

###**Series from a dict**

```python
info = {
    'name' : 'khushi',
    'age' : 22,
    'address' : 'Uttar Pradesh'
}

information = pd.Series(info)
information
```

##**Series methods**

```python
df = pd.read_csv('/content/placement.csv')
subs = df.iloc[:, 1]
subs.head() #default 5 rows
```

```python
subs.head(10)
```

```python
subs.tail() #default bottom 5rows
```

```python
subs.tail(2)
```

```python
subs.sample() #random 1 row
```

```python
subs.sample(4)
```

```python
#value counts

subs.value_counts() #counts of cgpa
```

```python
#sort values (not changing the original)
subs.sort_values() #default asc to desc
```

```python
subs.sort_values().head(1) #lowest cgpa
```

```python
#subs.sort_values(inplace=True)
#This Series is a view of some other array, to sort in-place you must create a copy

subs = df.iloc[:, 1].copy()
subs.sort_values(inplace=True)
subs
```

```python
#sort_index()
subs.sort_index()
```

##**Series indexing**

```python
#integer indexing
x = pd.Series([1,2,3,4,5,6,7,8,9])
x[4]
```

```python
#negative indexing
#x[-4] doesnt work
x[-3:] #works

#if the custom index is String, negative indexing wll work
```

```python
#slicing
df = pd.read_csv('/content/placement.csv')
subs = df.iloc[:, 2]
subs[2:4]
```

```python
#negative slicing
subs[-3:]
```

```python
#fancy indexing
subs[[1,3,5]]
```
